```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
data=pd.read_csv("heart_disease_uci.csv")
df=pd.DataFrame(data)
print(df.head(5))
```

```python
if 'id' in df.columns:
    df.drop('id', axis=1, inplace=True)
print(df.head(5))
```

```python
if 'dataset' in df.columns:
  df.drop('dataset', axis=1, inplace=True)
print(df.head(5))
```

```python
df.drop_duplicates(inplace=True)
```

```python
sns.pairplot(data=df)
plt.show()
```

```python
num_cols=['trestbps','chol','thalch','oldpeak','ca']
for col in num_cols:
  if col in df.columns:
    df[col]=df[col].fillna(df[col].median())
```

```python
cat_cols=['fbs','restecg','exang','slope','thal']
for col in cat_cols:
  if col in df.columns:
    df[col]=df[col].fillna(df[col].mode()[0])
    df[col]=df[col].astype(str)
```

```python
binary_cols=['fbs','restecg','exang','slope','thal','sex','cp']
df=pd.get_dummies(df,columns=binary_cols,prefix=binary_cols)
```

```python
from numpy import bool_
bool_cols=df.select_dtypes(include='bool').columns
df[bool_cols]=df[bool_cols].astype(int)
```

```python
df['target']=df['num'].apply(lambda x:1 if x>0 else 0)
df=df.drop(columns=['num'])
```

```python
x=df.drop(columns=['target'])
y=df['target']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x_train=scaler.fit_transform(x_train)
x_test=scaler.transform(x_test)
```

```python
print("Enter patient details for heart disease risk prediction:\n")

age = int(input("Age: "))
trestbps = float(input("Resting Blood Pressure (trestbps): "))
chol = float(input("Cholesterol (chol): "))
thalch = float(input("Max Heart Rate Achieved (thalch): "))
oldpeak = float(input("Oldpeak (ST depression): "))
ca = int(input("Number of major vessels colored (ca, 0-3): "))

fbs = input("Fasting Blood Sugar > 120 mg/dl? (True/False): ").strip()
restecg = input("Resting ECG (normal / lv hypertrophy / st-t abnormality): ").strip()
exang = input("Exercise Induced Angina? (True/False): ").strip()
slope = input("Slope (upsloping / flat / downsloping): ").strip()
thal = input("Thal (normal / fixed defect / reversable defect): ").strip()
sex = input("Sex (Male/Female): ").strip()
cp = input("Chest Pain Type (typical angina / atypical angina / non-anginal / asymptomatic): ").strip()


raw_input = {
    'age': age,
    'trestbps': trestbps,
    'chol': chol,
    'thalch': thalch,
    'oldpeak': oldpeak,
    'ca': ca,
}


raw_input[f'fbs_{fbs}'] = 1
raw_input[f'restecg_{restecg}'] = 1
raw_input[f'exang_{exang}'] = 1
raw_input[f'slope_{slope}'] = 1
raw_input[f'thal_{thal}'] = 1
raw_input[f'sex_{sex}'] = 1
raw_input[f'cp_{cp}'] = 1


input_df = pd.DataFrame([raw_input])
input_df=input_df.reindex(columns=x.columns,fill_value=0)
input_scaled=scaler.transform(input_df)
prediction = knn.predict(input_scaled)[0]
probability = knn.predict_proba(input_scaled)[0]


print("\n" + "="*40)
print("HEART DISEASE RISK PREDICTION")
print("="*40)
if prediction == 1:
    print("Result: AT RISK of heart disease")
else:
    print("Result: LOW RISK / No heart disease detected")

print(f"Confidence: {max(probability)*100:.2f}%")
print(f"Probability breakdown -> No Disease : {probability[0]*100:.2f}%, Disease : {probability[1]*100:.2f}%")
```

## Level 4: Prescriptive Analytics & Recommendations

### Recommendation 1: Priority Triage Rule
Patients flagged at-risk with ca ≥ 1 combined with elevated oldpeak should be
escalated for cardiologist review within 7 days, since these two features
show the strongest positive association with risk (r=0.52 and r=0.44).

### Recommendation 2: Stress-Test Screening Protocol
Since max heart rate (thalch) is negatively correlated with risk (r=-0.37),
clinics should flag patients over 50 showing below-median max heart rate
under stress testing for earlier intervention.

### Recommendation 3: Threshold / Trade-off Policy
Current model: Precision 87.2%, Recall 85.6%.

### Recommendation 4: Data Governance
Exclude non-clinical identifiers (id) from the feature set in future model
iterations, as they carry no legitimate predictive relationship with outcomes.
